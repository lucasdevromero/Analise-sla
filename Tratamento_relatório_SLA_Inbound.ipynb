```python
#Tratamento Fila
df_3_1.info()

# Alterando a coluna 'Documento' de float para object
df_3_1['Documento'] = df_3_1['Documento'].astype('object')

# Substituir valores NaN por uma string vazia
df_3_1['Tempo Fila'] = df_3_1['Tempo Fila'].fillna('')

# Converter para datetime (hora:minuto:segundo)
df_3_1['Tempo Fila'] = pd.to_datetime(df_3_1['Tempo Fila'], format='%H:%M:%S', errors='coerce').dt.time

# Convertendo a coluna para tipo datetime
df_3_1['Data Hora Inicio Operação'] = pd.to_datetime(df_3_1['Data Hora Inicio Operação'])
df_3_1['Data Hora Fim Operação'] = pd.to_datetime(df_3_1['Data Hora Fim Operação'])

# Corrigindo a fórmula de cálculo de Tempo Fila
def calcular_tempo_fila(row):
    # Verifica se Tempo Fila está vazio (NaN ou valor inválido)
    if pd.isna(row['Tempo Fila']) or not isinstance(row['Tempo Fila'], str):
        # Verifica se as colunas de Data Hora Início e Fim não estão vazias
        if pd.notna(row['Data Hora Inicio Operação']) and pd.notna(row['Data Hora Fim Operação']):
            # Calcula a diferença entre Data Hora Fim e Data Hora Início
            tempo_fila = row['Data Hora Fim Operação'] - row['Data Hora Inicio Operação']

            # Retorna a diferença no formato 'HH:MM:SS'
            return str(tempo_fila).split()[2]  # 'HH:MM:SS'
    return row['Tempo Fila']  # Retorna o valor já existente se for válido

# Aplicando a função para calcular o Tempo Fila
df_3_1['Tempo Fila'] = df_3_1.apply(calcular_tempo_fila, axis=1)

# Confirmando Alteração
#df_3_1.info()

display(df_3_1)
```

```python
#display(df_2_2)
df_2_2.info()
# Transformando INICIO CONFERÊNCIA em date time
df_2_2['INICIO CONFERÊNCIA'] = df_2_2['INICIO CONFERÊNCIA'].replace({
    'erro em ': '', 'não é data': ''
}, regex=False)

# Convertendo corretamente a data com barras
df_2_2['Data_e_Hora_Inicio'] = pd.to_datetime(
    df_2_2['INICIO CONFERÊNCIA'],
    format='%d/%m/%Y %H:%M:%S',
    errors='coerce'
)

# Verificando o resultado
df_2_2.info()
display(df_2_2)
```

```python
#Merge Controle Inbound

# Usando fillna() para substituir NaN na PLACA por uma string vazia, e depois concatenar
#df_2_2['Carga_Dia'] = df_2_2['DATA'] + '_' + df_2_2['PLACA'].fillna('')

df_2_2 = df_2_2.rename(columns={'DOCUMENTO': 'Documento'})


#Merge com o DF_merge
df_merge_final = pd.merge(df_2_2, df_merge, on='Documento')

#Merge com o df_3_1
df_merge_final = pd.merge(df_merge_final, df_3_1, on='Documento', how='left')

# Calculando a diferença entre 'Data_e_Hora_Fim' e 'Data_e_Hora_Inicio'
df_merge_final['Diferenca'] = df_merge_final['Data_e_Hora_Fim'] - df_merge_final['Data_e_Hora_Inicio']

# Calculando a diferença em horas (convertendo de timedelta para horas)
df_merge_final['Diferenca_em_Horas'] = df_merge_final['Diferenca'].dt.total_seconds() / 3600

import pandas as pd
import numpy as np
from datetime import timedelta

# Supondo que df_merge_final['DataHoraInicio'] seja a coluna com a data e hora de início
def calcular_sla(row):
    # Obter o dia da semana e a hora
    dia_da_semana = row['Data_e_Hora_Inicio'].weekday()  # 0 = segunda-feira, 4 = sexta-feira, 6 = domingo
    hora_inicio = row['Data_e_Hora_Inicio'].hour

    # Verificar se está entre sexta-feira após as 18h e domingo até as 23h
    if (dia_da_semana == 4 and hora_inicio >= 18) or (dia_da_semana == 5) or (dia_da_semana == 6 and hora_inicio <= 23):
        # Se estiver no intervalo, considera SLA de 72 horas
        limite_sla = timedelta(hours=72)
    else:
        # Caso contrário, considera SLA de 24 horas
        limite_sla = timedelta(hours=24)

    # Calcular a diferença de tempo
    diferenca = row['Diferenca_em_Horas']  # Supondo que já tenha a diferença de horas calculada em outra coluna

    # Definir o SLA Status
    if diferenca > limite_sla.total_seconds() / 3600:  # Convertendo limite_sla para horas
        return 'Fora do SLA'
    else:
        return 'Dentro do SLA'

# Aplicando a função para criar a coluna 'SLA_Status'
df_merge_final['SLA_Status'] = df_merge_final.apply(calcular_sla, axis=1)


# Tente converter as datas, transformando as inválidas em NaT (Not a Time)
df_merge_final['DATA'] = pd.to_datetime(df_merge_final['DATA'], format='%d/%m/%Y', errors='coerce')

#df_merge_final.info()

#display(df_merge_final)
```

```python
#gerando DF filtrado só com os fora do SLA

#definindo a coluna de finalização do turno

# Calculando o total de minutos a partir da coluna 'Data_e_Hora_Fim'
df_merge_final['Total_Minutos'] = df_merge_final['Data_e_Hora_Fim'].dt.hour * 60 + df_merge_final['Data_e_Hora_Fim'].dt.minute

# Lógica para determinar o turno
df_merge_final['Turno'] = df_merge_final['Data_e_Hora_Fim'].apply(lambda x: (
    "1º Turno" if x.weekday() == 4 and (x.hour * 60 + x.minute) >= 351 and (x.hour * 60 + x.minute) < 720 else
    "2º Turno" if x.weekday() == 4 and (x.hour * 60 + x.minute) >= 721 and (x.hour * 60 + x.minute) < 1080 else
    "3º Turno" if x.weekday() == 4 else
    "1º Turno" if (x.hour * 60 + x.minute) >= 351 and (x.hour * 60 + x.minute) < 860 else
    "2º Turno" if (x.hour * 60 + x.minute) >= 861 and (x.hour * 60 + x.minute) < 1360 else
    "3º Turno"
))

# Concatenando as colunas de data e placa em uma nova coluna
# Convertendo as colunas de data para string antes de concatenar
df_merge_final['Chave'] = df_merge_final['Documento'].astype(str) + '-' + \
                          df_merge_final['Data_e_Hora_Inicio_Fila'].astype(str) + '-' + \
                          df_merge_final['Data_e_Hora_Fim_Fila'].astype(str) + '-' + \
                          df_merge_final['Status'].astype(str) + '-' + \
                          df_merge_final['Tempo Fila'].astype(str) + '-' + \
                          df_merge_final['Data_Agenda'].astype(str) + '-' + \
                          df_merge_final['PLACA'].astype(str) + '-' + \
                          df_merge_final['DOCA'].astype(str) + '-' + \
                          df_merge_final['STATUS'].astype(str) + '-' + \
                          df_merge_final['TIPO DE CARGA'].astype(str) + '-' + \
                          df_merge_final['Data_e_Hora_Inicio'].dt.strftime('%Y-%m-%d %H:%M:%S') + '-' + \
                          df_merge_final['Data_e_Hora_Fim'].dt.strftime('%Y-%m-%d %H:%M:%S') + '-' + \
                          df_merge_final['Diferenca'].astype(str) + '-' + \
                          df_merge_final['SLA_Status_Conf_Arma'].astype(str)



#display(df_merge_final)
#display(df_fora_do_SLA)
```
